```python
#E_inv = np.array(((1, -2), (-2, 1)))
#E_inv = np.array(((-1/3, -2/3), (2/3, -1/3)))

#print(E_inv)
def pdf(x, E, E_inv):
    #k=2, det=1
    #print((x[:,:,None,:] @ E_inv[None,:]).shape)
    # (2 * np.pi)**-1 *
    #return  np.exp(-1/2 * x[:,:,None,:] @ E_inv[None,:] @ x[:,:,:,None])
    std_norm = np.log(np.linalg.det(E)**(-0.5) * (2 * np.pi)**-1)
    return std_norm + -1/2 * x[:,:,None,:] @ E_inv[None,:] @ x[:,:,:,None]
#pdf(mat_coords)
```

```python
### size = 20
x = np.arange(-size, size, 0.5)
y = np.arange(-size, size, 0.5)
xx, yy = np.meshgrid(x, y)

mat_coords = np.stack((xx,yy), axis=-1)

E = np.array(((3,0), (0, 4)))
E_inv = np.linalg.inv(E)
print(np.linalg.det(E))
pdfs = pdf(mat_coords, E, E_inv)
print(pdfs.shape)

pdfs = pdfs.squeeze()
print(pdfs[:,:].shape)
plt.contourf(xx, yy, pdfs.squeeze())#samples_2d[:,0], samples_2d[:,1], c=np.log(pdf), cmap='rainbow', alpha=0.2)
```
