```python
# normal_curve.py

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
# if using a Jupyter notebook, inlcude:
%matplotlib inline
```

```python
# define constants
mu = 0
sigma = 1

offset = 2.1
x1 = -offset
x2 = offset

x = np.arange(x1, x2, 0.001) # range of x in spec
x_left = np.arange(-10, x1, 0.001) # entire range of x, both in and out of spec
x_right = np.arange(x2, 10, 0.001) # entire range of x, both in and out of spec

# mean = 0, stddev = 1, since Z-transform was calculated
y = norm.pdf(x,0,1)
y_left = norm.pdf(x_left,0,1)
y_right = norm.pdf(x_right,0,1)
```

```python
# build the plot
fig, ax = plt.subplots(figsize=(9,6))
plt.style.use('fivethirtyeight')
ax.plot(x,y, c='black', linewidth=1)
ax.plot(x_left,y_left, c='black', linewidth=1)
ax.plot(x_right,y_right, c='black', linewidth=1)


alph = 0.9
ax.fill_between(x_left,y_left,0, alpha=alph, color='r')
ax.fill_between(x_right,y_right,0, alpha=alph, color='r')

ax.fill_between(x,y,0, alpha=0.7, color='b')

ax.set_xlim([-4,4])
ax.set_xlabel('p(x)')
#ax.set_xlabel('Normal Distribution - Red Areas are outliers')
#ax.set_xlabel('# of Standard Deviations Outside the Mean')
ax.set_yticklabels([])
#ax.set_title('Normal Gaussian Curve')

plt.savefig('normal_curve.png', dpi=72, bbox_inches='tight')
plt.show()
```
